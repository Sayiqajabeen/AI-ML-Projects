````python
import matplotlib.pyplot as plt

months     = ['Jan','Feb','Mar','Apr','May','Jun']
sales      = [200, 240, 180, 300, 350, 320]

cities     = ['Karachi','Lahore','Islamabad','Peshawar']
population = [16, 13, 2, 4]   # millions

hours      = [1, 2, 3, 4, 5, 6, 7, 8]
grades     = [50, 55, 60, 65, 72, 78, 85, 90]

categories = ['Food', 'Rent', 'Transport', 'Entertainment']
expenses   = [40, 30, 15, 15]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Top-left: line plot of sales over months
axes[0,0].plot(months, sales, marker='o')
axes[0,0].set_title("Sales Over Months")

# Top-right: bar chart of city populations
axes[0,1].bar(cities, population)
axes[0,1].set_title("City Population (Millions)")

# Bottom-left: scatter plot of hours vs grades
axes[1,0].scatter(hours, grades)
axes[1,0].set_title("Hours vs Grades")

# Bottom-right: pie chart of expenses

axes[1,1].pie(expenses, labels=categories, autopct='%1.0f%%')
axes[1,1].set_title("Monthly Expenses")

fig.suptitle("Monthly Dashboard", fontsize=14)
plt.tight_layout()
plt.show()
````

````python
# from bokeh.plotting import figure, show
# from bokeh.io import output_notebook
# from bokeh.models import HoverTool, ColumnDataSource

# output_notebook()

# days     = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
# visitors = [120, 180, 150, 200, 250, 300, 270]

# # YOUR CODE HERE
# # 1. Create a ColumnDataSource
# source = ColumnDataSource(data=dict(days=days, visitors=visitors))  

# # 2. Create a Bokeh figure with title and axis labels
# p = figure(title="Daily Visitors", x_axis_label='Days', y_axis_label='Number of Visitors')


# # 3. Add a line renderer;lines connecting data points.
# p.line('days', 'visitors', source=source, line_width=2,color='blue', legend_label="Visitors")

# # 4. Add circle markers
# p=figure(title="Daily Visitors", x_axis_label='Days', y_axis_label='Number of Visitors')
# p.line('days', 'visitors', source=source, line_width=2)

# # 5. Add a HoverTool
# hover = HoverTool(tooltips=[("Days", "days"), ("Visitors", "@visitors")])
# p.add_tools(hover)


# # 6. Show the plot
# show(p)  

from bokeh.plotting import figure, show
from bokeh.io import output_notebook
from bokeh.models import HoverTool, ColumnDataSource

output_notebook()

# Data
days     = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
visitors = [120, 180, 150, 200, 250, 300, 270]

# 1. ColumnDataSource
source = ColumnDataSource(data=dict(day=days, visitors=visitors))

# 2. Create figure (use width & height instead of plot_width/plot_height)
p = figure(title="Website Visitors Over a Week",
           x_range=days,
           y_axis_label="Number of Visitors",
           x_axis_label="Day",
           width=700,
           height=400)

# 3. Add line renderer
p.line('day', 'visitors', source=source, line_width=2, color='blue', legend_label="Visitors")

# 4. Add circle markers
p.circle('day', 'visitors', source=source, size=8, color='red', legend_label="Visitors")

# 5. Add HoverTool
hover = HoverTool(tooltips=[("Day", "@day"), ("Visitors", "@visitors")])
p.add_tools(hover)

# 6. Show the plot
show(p)
````

## 🤔 Reflection Questions

Answer the following in the code cell below as comments:

1. What is the main advantage of using **Seaborn** over raw Matplotlib for statistical plots?
2. When would you choose **Bokeh** over **Plotly**, and vice versa?
3. In Task 3, how does changing the number of bins affect your interpretation of the distribution?
4. Why is a **pie chart** generally not recommended when you have more than 5–6 categories?
5. What does a **violin plot** show that a **box plot** does not?


````python
# # Q1:
# 1. Main advantage of Seaborn over raw Matplotlib

# Seaborn is built on top of Matplotlib and is designed for statistical visualization.

# Advantages:

# Simplifies complex plots like box plots, violin plots, pair plots, and heatmaps with fewer lines of code.

# Automatically handles color palettes, themes, and styling.

# Integrates directly with Pandas DataFrames, so you can plot columns without converting to arrays.

# Example: A single line of Seaborn code can produce a styled box plot with grouping, whereas Matplotlib requires multiple lines and manual customization.





# Q2:2. When to choose Bokeh vs Plotly
# Library	When to Use
# Bokeh	     -   For large, interactive web dashboards.
#            - Works well with real-time streaming or updating plots.
#            - Integrates nicely with Bokeh server for interactive applications.
# Plotly   	- For interactive charts that work in notebooks or web apps.
#            - Easy for animated plots, bubble charts, and dashboards without a server.
#            - Better for quick interactive visualization in Jupyter notebooks.

# Rule of thumb:

# Bokeh = more control for live, web-ready dashboards.

# Plotly = easier for interactive exploratory analysis.



# Q3:3. Effect of changing the number of bins (Task 3 — Histogram)

# The number of bins determines how the data is grouped along the x-axis.

# Fewer bins → broader groups, may hide fine details, smoother histogram.

# More bins → narrower groups, may show more detail but can be noisy.

# Choosing the right number of bins affects how you interpret the distribution: too few hides patterns, too many may overemphasize minor variations.





# Q4: 4. Why pie charts are not recommended for >5–6 categories

# Pie charts become hard to read when slices are too many.

# Humans have difficulty comparing angles, especially small slices.

# Colors may repeat or become confusing, reducing clarity.

# Alternatives: bar charts or stacked bar charts are easier to interpret for many categories.




# Q5:
# . What a violin plot shows that a box plot does not

# A violin plot combines a box plot with a kernel density estimate (KDE).

# While a box plot shows median, quartiles, and outliers, a violin plot also shows:

# The full distribution shape of the data.

# Multiple peaks or skewness.

# Example: If the data is bimodal, the box plot may hide it, but the violin plot will reveal both peaks.

````
