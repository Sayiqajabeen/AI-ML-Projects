## ⭐ Task 3 — Quartiles & Five-Number Summary

**Dataset:** Marks of 12 students:
```python
marks = [45, 52, 58, 61, 67, 72, 76, 80, 85, 91, 95, 99]
```

**Your job:**
1. Find **Q1**, **Q2 (median)**, **Q3**
2. Calculate the **IQR** (Q3 − Q1)
3. Print the full **five-number summary** (Min, Q1, Q2, Q3, Max)

**Expected output:**
```
=== Five-Number Summary ===
Min    : 45
Q1     : 60.25
Median : 74.0
Q3     : 85.75
Max    : 99
IQR    : 25.5
```

> 💡 **Hints:**
> - `np.percentile(data, 25)` → Q1
> - `np.percentile(data, 50)` → Q2
> - `np.percentile(data, 75)` → Q3
> - IQR = Q3 − Q1

````python
# Task 3 — Your code here
marks = [45, 52, 58, 61, 67, 72, 76, 80, 85, 91, 95, 99]

# Q1, Q2, Q3

q1 = np.percentile(marks, 25)
q2 = np.percentile(marks, 50)
q3 = np.percentile(marks, 75)
iqr = q3 - q1
# IQR
quartile_deviation = (q3 - q1) / 2
print("Interquartile Range (IQR):", iqr)
print("Quartile Deviation:", quartile_deviation)

# Five-number summary
print('=== Five-Number Summary ===')
print(f'Min    : {min(marks)}')
print(f'Q1     : {q1}')
print(f'Median : {q2}')
print(f'Q3     : {q3}')
print(f'Max    : {max(marks)}')
print(f'IQR    : {iqr}')

low_fence  = q1 - 1.5 * iqr
high_fence = q3 + 1.5 * iqr
print(f'\nOutlier fences: {low_fence:.1f}  to  {high_fence:.1f}')
outliers = [x for x in marks if x < low_fence or x > high_fence]
print(f'Outliers found: {outliers if outliers else "None"}')
````

## ⭐⭐ Task 5 — Z-Score Normalisation

**Dataset:** Heights (cm) of 10 people:
```python
heights = [150, 162, 170, 155, 180, 175, 168, 158, 172, 165]
```

**Your job:**
1. Compute the **mean** and **std** of heights
2. Calculate the **z-score** for each height
3. Print each height with its z-score
4. Flag anyone with **|z| > 1.5** as `'⚠️ Unusual'`

**Expected output:**
```
Height  Z-Score  Flag
 150     -1.84   ⚠️ Unusual
 162     -0.68
 170     +0.09
 ...
 180     +1.08
```

> 💡 **Hints:**
> - `z = (x - mean) / std`
> - Loop: `for h in heights:`
> - `flag = '⚠️ Unusual' if abs(z) > 1.5 else ''`
> - This is exactly what `StandardScaler` does in scikit-learn!

````python
# Task 5 — Your code here
heights = [150, 162, 170, 155, 180, 175, 168, 158, 172, 165]

# Compute mean and std
mean = np.mean(heights)
std = np.std(heights)
# Print table header
print(f'  {"Height":>6}  {"Z-Score":>8}  Flag')
print('-' * 30)

# Calculate and print z-scores
for h in heights:
    z =  (h - mean) / std
    if abs(z) > 2:
        flag = "Unusual"
    else:
        flag = "Normal"
    print(f'{h:>8}  {z:>8.2f}  {flag}')
   
````

````python
# Task 6 — Your code here
from turtle import mode


from statistics import median, mode, mean 


right_skewed = [2, 3, 3, 4, 4, 4, 5, 5, 10, 15, 20]
left_skewed  = [1, 5, 10, 15, 16, 16, 17, 17, 18, 18, 19]

# For right_skewed:
#Large values pull the mean to the right.
right_skewed.sort()
print("Right-skewed data (sorted):", right_skewed)      

r_mean = mean(right_skewed)
r_median = median(right_skewed)
r_mode = mode(right_skewed)

print("\nRight-skewed statistics")
print("Mean:", r_mean)
print("Median:", r_median)
print("Mode:", r_mode)

# For left_skewed:
#Small values pull the mean to the left.
left_skewed.sort()
print("Left-skewed data (sorted):", left_skewed)  
l_mean = mean(left_skewed)
l_median = median(left_skewed)
l_mode = mode(left_skewed)

print("\nLeft-skewed statistics")
print("Mean:", l_mean)
print("Median:", l_median)
print("Mode:", l_mode)



# Print and verify the skew rule
#Right skew: Mean > Median > Mode
#Left skew: Mean < Median < Mode
print("\nSkewness Verification")

if r_mean > r_median > r_mode:
    print("Right-skewed rule verified: Mean > Median > Mode")

if l_mean < l_median < l_mode:
    print("Left-skewed rule verified: Mean < Median < Mode") 
````

## ⭐⭐ Task 7 — Correlation Analysis

**Dataset:** Hours of sleep and test scores for 10 students:
```python
hours_sleep = [5, 6, 6, 7, 7, 8, 8, 9, 9, 10]
test_score  = [55, 60, 62, 68, 71, 75, 78, 82, 85, 88]
```

**Your job:**
1. Calculate the **Pearson correlation** and **p-value**
2. Determine strength: Strong (>0.8) / Moderate (>0.5) / Weak
3. State if the result is statistically significant (p < 0.05)
4. **Bonus:** Plot a scatter plot with a trend line

**Expected output:**
```
Pearson r = 0.997
p-value   = 0.000001
Strength : STRONG positive correlation
Significant: YES ✅
```

> 💡 **Hints:**
> - `corr, pval = stats.pearsonr(x, y)`
> - Use `if/elif/else` for strength labels
> - Bonus: `plt.scatter(x, y)` then `np.polyfit(x, y, 1)` for trend line

````python
# Task 7 — Your code here
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

hours_sleep = [5, 6, 6, 7, 7, 8, 8, 9, 9, 10]
test_score  = [55, 60, 62, 68, 71, 75, 78, 82, 85, 88]

# 1. Pearson correlation
r, p_value = pearsonr(hours_sleep, test_score)

print("Pearson correlation (r):", round(r,3))
print("p-value:", p_value)
# 2. Strength classification
abs_r = abs(r)

if abs_r >= 0.8:
    strength = "Very Strong"
elif abs_r >= 0.6:
    strength = "Strong"
elif abs_r >= 0.4:
    strength = "Moderate"
elif abs_r >= 0.2:
    strength = "Weak"
else:
    strength = "Very Weak"

print("Correlation strength:", strength)

# 3. Significance check
if p_value < 0.05:
    print("Result is statistically significant")
else:
    print("Result is NOT statistically significant")

# 4. (Bonus) Scatter plot
plt.scatter(hours_sleep, test_score)

plt.title("Sleep vs Test Score")
plt.xlabel("Hours of Sleep")
plt.ylabel("Test Score")

plt.show()
````

## ⭐⭐⭐ Task 9 — Hypothesis Testing

**Scenario:** A school tested a new teaching method.

```python
old_method = [65, 70, 68, 72, 74, 69, 71, 73, 66, 75]
new_method = [78, 82, 80, 85, 79, 83, 81, 84, 77, 86]
```

**Your job:**
1. State **H₀** and **H₁** in plain English (as print statements)
2. Run an **independent samples t-test**
3. Print t-statistic and p-value
4. Print the means of both groups
5. Write a clear **conclusion** based on the p-value

**Expected output:**
```
H₀: Both methods produce the same results
H₁: The new method is significantly different

Old method mean: 70.3
New method mean: 81.5
t-statistic    : -10.470
p-value        : 0.000000

✅ Reject H₀ — new method is significantly better!
```

> 💡 **Hints:**
> - `stats.ttest_ind(group1, group2)`
> - If p < 0.05 → reject H₀
> - Compare the means to see which group performed better
> - The t-stat being negative just means group1 < group2

````python
# Task 9 — Your code here
import scipy.stats as stats
import numpy as np
old_method = [65, 70, 68, 72, 74, 69, 71, 73, 66, 75]
new_method = [78, 82, 80, 85, 79, 83, 81, 84, 77, 86]

# 1. State hypotheses
print("H₀: Both methods produce the same results")
print("H₁: The new method produces significantly different results\n")
# 2. Run t-test
t_stat, p_value = stats.ttest_ind(old_method, new_method)
# 3. Print t-stat and p-value
print(f"t-statistic    : {t_stat:.3f}")
print(f"p-value        : {p_value:.6f}")
# 4. Print group means
old_mean = np.mean(old_method)
new_mean = np.mean(new_method)
print(f"\nOld method mean: {old_mean:.1f}")
print(f"New method mean: {new_mean:.1f}")

# 5. Conclusion
if p_value < 0.05:
    print("\n✅ Reject H₀ — new method is significantly better!")
else:
    print("\n❌ Fail to reject H₀ — no significant difference.")
````

````python
import numpy as np
from scipy import stats

def stats_report(data, name):
    data = np.array(data)
    
    print('=' * 40)
    print(f'STATS REPORT: {name}')
    print('=' * 40)
    
    # 1. Values
    print(f'Values    : {list(data)}\n')
    
    # 2. Central Tendency
    mean = np.mean(data)
    median = np.median(data)
    
    mode_res = stats.mode(data, keepdims=True)
    mode_val = mode_res.mode[0]
    mode_count = mode_res.count[0]
    
    print('--- Central Tendency ---')
    print(f'Mean      : {mean:.2f}')
    print(f'Median    : {median:.2f}')
    print(f'Mode      : {mode_val} (appeared {mode_count}x)\n')
    
    # 3. Spread
    std = np.std(data, ddof=1)
    var = np.var(data, ddof=1)
    
    print('--- Spread ---')
    print(f'Std Dev   : {std:.2f}')
    print(f'Variance  : {var:.2f}\n')
    
    # 4. Quartiles
    Q1 = np.percentile(data, 25)
    Q2 = np.percentile(data, 50)
    Q3 = np.percentile(data, 75)
    IQR = Q3 - Q1
    
    print('--- Quartiles ---')
    print(f'Q1={Q1:.2f}  Q2={Q2:.2f}  Q3={Q3:.2f}  IQR={IQR:.2f}\n')
    
    # 5. Skewness
    skewness = stats.skew(data)
    direction = "right skewed" if skewness > 0 else "left skewed"
    
    print('--- Shape ---')
    print(f'Skewness  : {skewness:.2f} ({direction})\n')
    
    # 6. Outliers (IQR method)
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    print('--- Outliers (IQR method) ---')
    outliers = data[(data < lower) | (data > upper)]
    
    if len(outliers) == 0:
        print('No outliers found')
    else:
        for val in outliers:
            print(f'🚨 Outlier: {val}')
    
    # 7. Z-scores
    print('\n--- Z-Scores ---')
    z_scores = stats.zscore(data)
    
    for val, z in zip(data, z_scores):
        if abs(z) > 2:
            print(f'{val:>5}   z={z:+.2f} ⚠️ EXTREME')
        else:
            print(f'{val:>5}   z={z:+.2f}')
    
    print('=' * 40)


# Test it!
stats_report([12, 15, 14, 10, 100, 13, 14, 15, 16, 11], 'Sample Data')
````
