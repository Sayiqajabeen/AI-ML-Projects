Square root of real or complex number

```python
#Find square root of real or complex numbers  
#Importing the complex math module 
import cmath 
num=2+5j

#num=evl(input('Enter a number')) 

num_sqrt = cmath.sqrt(num)
print('The sqaure root of {0}, {1:0.3f}+{2:0.3f}j'.format(num, num_sqrt.real, num_sqrt.imag))
```

Convert Temprature celsius to fahrenheit

```python
celsius=16
#Now for fahrenheit 
fahrenheit=(celsius * 1.8)+32
print('%0.1f degree celsius is equal to %0.1f degree  fahrenheit'%(celsius, fahrenheit))
```

Area of Triangle

```python
# a=13
# b=14
# c=15

a=float(input("Enter fist side;"))
b=float(input("Enter second side;"))
c=float(input("Enter third side;"))

#calculate semi-parameter 
s=(a+b+c)/2 
#calcualte area
area=(s*(s-a)*(s-b)*(s-c))**0.5
print("The area of triangle is %0.2f"%area) 

```

Quadratic Formula/Equation ax**2+bx+c=0 

```python
#import math complex module 
import cmath
a=2
b=6
c=8

#Now calculate discriminant
d=(b**2)-(4*a*c) 


#Find two solutions
solution1=(-b-cmath.sqrt(d))/(2*a)
solution2=(-b+cmath.sqrt(d))/(2*a)  


print('The solution are {0} and  {1}'.format(solution1,solution2))
```

Kilometer to miles

```python
kilometers = float(input("Enter value in kilometer: "))

conv_fac = 0.621371
miles = kilometers * conv_fac

print(f"{kilometers} kilometers is equal to {miles:.2f} miles")
```
