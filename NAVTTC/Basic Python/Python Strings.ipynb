##### Python Strings  
In Python, a string is a sequence of characters. For example, "hello" is a string containing a sequence of characters 'h', 'e', 'l', 'l', and 'o'.

We use single quotes or double quotes to represent a string in Python.

```python
# create a string using double quotes
string1 = "Class of NIAI"

# create a string using single quotes
string1 = 'Class of NIAI'
```

#### Access String Characters in Python
We can access the characters in a string in three ways.

- **Indexing:** One way is to treat strings as a list and use index values. 

```python
greet = 'Lahore'

# access 1st index element
print(greet[1]) # "e"
```

- **Negative Indexing:** Similar to a list, Python allows negative indexing for its strings. 

```python
greet = 'Sayiqa'
# access 4th last element
print(greet[-4]) # "e"
```

- **Slicing:** Access a range of characters in a string by using the slicing operator colon : .

```python
greet = 'Sayiqa'

# access character from 1st index to 3rd index
print(greet[1:4])  # "ell"
```

#### Python Strings are Immutable
In Python, strings are immutable. That means the characters of a string cannot be changed. For example,

```python
message = 'NAVTTAC AI classes'
message[0] = 'H'
print(message)
```

However, we can assign the variable name to a new string

```python
message = 'Hello Students'

# assign new string to message variable
message = 'welcome to AI CLass'

print(message); # prints "Hello Friends"
```

#### Python Multiline String
We can also create a multiline string in Python. For this, we use triple double quotes """ or triple single quotes '''. 

```python
# multiline string 
message = """
Never gonna give you up
Never gonna let you down
"""
print(message)
```

#### Python String Operations
Many operations can be performed with strings, which makes it one of the most used data types in Python.

**1. Compare Two Strings** 

We use the == operator to compare two strings. If two strings are equal, the operator returns True. Otherwise, it returns False. For example,

```python
str1 = "Welcome to NIAI NATTAC Class"
str2 = "Im a Programmer."
str3 = "Welcome to NIAI NATTAC Class"

# compare str1 and str2
print(str1 == str2)

# compare str1 and str3
print(str1 == str3)
```

**2. Join Two or More Strings**

In Python, we can join (concatenate) two or more strings using the + operator.

```python
greet = "Hello, "
name = "Students"

# using + operator
result = greet + name
print(result)

# Output: Hello, Jack
```

#### Iterate Through a Python String
We can iterate through a string using a for loop

```python
greet = 'Hello'

# iterating through greet string
for letter in greet:
    print(letter)
```

#### Python String Length
In Python, we use the len() method to find the length of a string.

```python
name = 'Sayiqa'

# count length of greet string
print(len(name))

# Output: 5
```

#### String Membership Test
We can test if a substring exists within a string or not, using the keyword in.

```python
print('g' in 'program') 
print('t' not in 'battle') 
```

#### Escape Sequences in Python
The escape sequence is used to escape some of the characters present inside a string.

Suppose we need to include both a double quote and a single quote inside a string

```python
example = "He said, "What's there?"

print(example) # throws error
```

Since strings are represented by single or double quotes, the compiler will treat "He said, " as a string. Hence, the above code will cause an error.

To solve this issue, we use the escape character \ in Python.

```python
# escape double quotes
example = "He said, \"What's there?\""

# escape single quotes
example = 'He said, "What\'s there?"'

print(example)

# Output: He said, "What's there?"
```
